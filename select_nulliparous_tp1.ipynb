```r
double_pregnancy <- read.csv("/ceph/projects/010_SweMaMi/bm.nowak/data/2025-07-29_double_pregnancy.csv", header = TRUE)
head(double_pregnancy)
dim(double_pregnancy)
```

```r
# Convert char Q1_Datum to date format
double_pregnancy$Q1_Datum <- as.Date(double_pregnancy$Q1_Datum)
```

```r
# Sort file; group duplicate Q1_Personnummer by date
sorted_double_pregnancy <- double_pregnancy |>
  arrange(Q1_Personnummer, Q1_Datum)
glimpse(sorted_double_pregnancy)
```

```r
# Extract first and second pregnancy
kit_fp <- sorted_double_pregnancy |>
  group_by(Q1_Personnummer) |>
  slice(1)
```

```r
kit_sp <- sorted_double_pregnancy |>
  group_by(Q1_Personnummer) |>
  slice(2)
```

```r
# Check that num of rows and columns is still correct
dim(kit_fp)
dim(kit_sp)
```

```r
# Check dates and pers# are correct
head(kit_fp)
head(kit_sp)
```

```r
# write to file
write.table(kit_fp, file = paste0("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/saliva/tp1_nulliparous/", Sys.Date(), "_kit_fp_", ".csv"), sep = ",", row.names = FALSE)
write.table(kit_sp, file = paste0("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/saliva/tp1_nulliparous/", Sys.Date(), "_kit_sp_", ".csv"), sep = ",", row.names = FALSE)
```

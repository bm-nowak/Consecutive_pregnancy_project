# Load Files ----------------------------------------------------------------------------------------------------------

```r
# Read taxa data
taxa_tab <- read.table("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/vaginal/2025-11-11_vaginal_taxa_filtered.tsv", header = TRUE, sep ="\t", check.names = FALSE, quote='"') 
dim(taxa_tab)
head(taxa_tab)
```

```r
# Read metadata
combined_meta_tp1 <- read.csv("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/vaginal/tp1_nulliparous/2025-11-11_combined_tp1_filt_nulliparous.csv", header = TRUE, row.names = 1)
head(combined_meta_tp1)
dim(combined_meta_tp1)
```

# Sort taxa table ---------------------------------------------------------------------------------------

```r
head(taxa_tab)
```

```r
# Flipping the table 
taxa_flipped <- taxa_tab |> t() |>
    as.data.frame() |>
    rownames_to_column("sample_id")

taxa_flipped |> head()
```

```r
# Extract sample id, personal num, groups
```

```r
grep("kit1", names(combined_meta_tp1), value = TRUE)
```

```r
variables_df <- combined_meta_tp1 |> select(Group, Studienummer, Q1_Personnummer, 
                                            kit1.vaginal_sample.barcode, BMI_prior) 
head(variables_df)
```

```r
meta_BMI_update <- variables_df %>%
 mutate(
  BMI_group_prior = case_when(
   BMI_prior < 18.5 ~ "Underweight",
   BMI_prior >= 18.5 & BMI_prior <= 24.9 ~ "Normalweight",
   BMI_prior >= 25 & BMI_prior <= 29.9 ~ "Overweight",
   BMI_prior >= 30 ~ "Obese",
   TRUE ~ NA_character_  
 ))

head(meta_BMI_update$BMI_group_prior)
head(meta_BMI_update )
```

```r
# sort df so same paricipants grouped together
variables_df_sorted <- meta_BMI_update |>  sort_by(meta_BMI_update$Q1_Personnummer)
head(variables_df_sorted)
```

```r
# Order taxa table accoring to metadata
```

```r
# Remove rownames
variables_df_sorted <- variables_df_sorted |> rownames_to_column ("old") |> 
    select(- old, -BMI_prior) |> 
    column_to_rownames(var = "kit1.vaginal_sample.barcode") |>
    rownames_to_column("sample_id")
head(variables_df_sorted)
```

```r
# Now sort the taxa table according to the above metadata 
taxa_flipped_sorted <- taxa_flipped[match(variables_df_sorted$sample_id, taxa_flipped$sample_id ),]
taxa_flipped_sorted |> head()
```

```r
# Remove rownames
taxa_flipped_sorted <- taxa_flipped_sorted |> rownames_to_column ("old") |> 
    select(- old) 
head(taxa_flipped_sorted)
```

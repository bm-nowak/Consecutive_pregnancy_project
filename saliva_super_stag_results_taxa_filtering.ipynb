# Joining df -------------------------------------------------------------------------------------------------------------------------

 Files are merged horizontaly (each file added to the right).
- While adding, check that each row matches and if there are new rows then add new row.
- full_join merges by keeping rows from both and adding NA if no match.
- Join by using species name (clade) 

```r
# matches first "s__", then any char after it ("+"),  may end "([...])?" with "_unclassified", 
# match goes to end of str "$"
match_pattern <- "s__.+(_unclassified)?$" 
```

```r
# Reads each file in setwd if file name present in input list
# Also only keeps rows that match pattern
f_join_tables <- function(input_list_of_files){
    lapply(input_list_of_files, function(a_file){
        
        files_tabled <- read.table(a_file, header = TRUE)
        
        #cat("\nProcessing:", a_file, "\n") # debug line
        print(dim(files_tabled)) # Dim for each df
        
        # finds rows where pattern TRUE (excludes unnecessary taxa info etc since pattern match)
        filtered_tab <- files_tabled |> filter(grepl(match_pattern, clade_name)) 
                
        return(filtered_tab)
    })
    }

all_df_list <- f_join_tables(file_list)
head(all_df_list[[1]]) # check first data frame
```

```r
head(all_df_list[[32]]) # check last data frame
```

```r
# full_join merges by keeping rows from both and adding NA if no match and join by using species name (clade) 

joined_table <- reduce(all_df_list, full_join, by = "clade_name") 
# reduce takes each df and joins them together with the previous ones, 
# I.e instead of looping: append this_df with that_df, then append this_df with that_df_2 etc, it takes resulting df and full joins the next df in the list to it

#head(joined_table)
dim(joined_table)
tail(joined_table)
#colnames(joined_table)
```

# Minimum abundance and Prevalence Filtering --------------------------------------------------------------------------------------------------------

Left to do:
- Replace NA with 0
- Clean up strains
- Remove Samples with colSums < 80
- Add other taxa

```r
min_filtered_joined_table <- joined_table %>% {replace(., is.na(.), 0)}
tail(min_filtered_joined_table)
```

```r
# Rearrange to keep only the species name
sp <- min_filtered_joined_table$clade_name # Extracts clade names
aa <- sub('.*s__', '', sp) # Remove all characters before s__ ("sub()" replaces ".*s__" any char (including s__) with empty string "''" ) 
# eg s__Corynebacterium_pseudogenitalium --> becomes Corynebacterium_pseudogenitalium

#head(min_filtered_joined_table)

t_strain <- grep("t__", aa, value = TRUE) # finds clade-name containing t__ 
#t_strain # list of clades containing "t__"

min_filtered_joined_table$clade_name <- aa # replaces clade name with cleaned species name
new_filtered_tab <- min_filtered_joined_table[!min_filtered_joined_table$clade_name %in% t_strain,] # If a row name (clade_name) appears in list containing "t__" then remove it
```

```r
head(new_filtered_tab)
```

```r
# Function for removing minimum abundance and prevalence taxa
is_here <- function(invector, min_abund, min_pres){
  min_count <- length(invector) * min_pres
  these <- length(which(invector > min_abund))
  if(these > min_count){
    return(TRUE)
  }
  else{
    return(FALSE)
  }
}

keep <- apply(new_filtered_tab , MARGIN = 1, is_here, min_abund = 0.001, min_pres = 0.05)
filt_taxa <- new_filtered_tab [keep,]
head(filt_taxa)
dim(filt_taxa) 
```

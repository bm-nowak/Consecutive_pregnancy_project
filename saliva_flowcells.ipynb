```r
library(tidyverse)
library(dplyr)
```

# Read files =================================================================================================

```r
combined_metadata <- read.csv("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/saliva/tp1_nulliparous/2025-10-16combined_for_flowcells_saliva.csv", header = TRUE, row.names = 1)
head(combined_metadata)
dim(combined_metadata)
```

```r
saliva_samples <- read.csv("/ceph/projects/010_SweMaMi/bm.nowak/data/saliva_samples_250923.csv", header = TRUE, row.names = 1)
head(saliva_samples)
dim(saliva_samples)
```

# TP1----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------

```r
grep("kit1", names(combined_metadata), value = TRUE)
```

```r
head(combined_metadata$kit1.salival_sample.barcode)
```

```r
# Find sample ids that are valid
sample_ids_to_pull <- combined_metadata |>
    pull(kit1.salival_sample.barcode)
length(sample_ids_to_pull)
head(sample_ids_to_pull)
```

```r
# Pull only valid sample ids
flowcell_ids <- saliva_samples |>
    filter(sample_id %in% sample_ids_to_pull) |>
    pull(flowcell_id)
length(flowcell_ids)
```

```r
head(flowcell_ids)
```

```r
class(flowcell_ids)
```

# TP2----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------

```r
# Find sample ids that are valid for kit2 in 
sample_ids_to_pull_kit2 <- combined_metadata |>
    filter(!(is.na(kit2.salival_sample.barcode))) |>
    pull(kit2.salival_sample.barcode)
length(sample_ids_to_pull_kit2)
head(sample_ids_to_pull_kit2)
```

```r
# Pull only valid sample ids
flowcell_ids_kit2 <- saliva_samples |>
    filter(sample_id %in% sample_ids_to_pull_kit2) |>
    pull(flowcell_id)
length(flowcell_ids_kit2)
```

# Write to file ------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------

```r
writeLines(flowcell_ids, file.path("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/saliva/flowcells/", paste0(Sys.Date(), "_flowcell_ids_kit1.txt")))
```

```r
writeLines(flowcell_ids_kit2, file.path("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/saliva/flowcells/", paste0(Sys.Date(), "_flowcell_ids_kit2.txt")))
```

```r
all_flow_cell_id <- c(flowcell_ids , flowcell_ids_kit2)
flowcell_id_uni <- unique(all_flow_cell_id )
```

```r
# Write a combined file
writeLines(flowcell_id_uni, file.path("/ceph/projects/010_SweMaMi/bm.nowak/filtered_data/saliva/flowcells/", paste0(Sys.Date(), "flowcell_id_uni.txt")))
```
